```python

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Step 1: Get the top 8 and bottom 8 countries
top8 = df.nlargest(8, 'Ladder score').copy()
top8['Group'] = 'Happiest 8'

bottom8 = df.nsmallest(8, 'Ladder score').copy()
bottom8['Group'] = 'Least Happy 8'

# Combine all 16 countries
combined = pd.concat([bottom8, top8], ignore_index=True)
combined = combined.sort_values('Ladder score', ascending=True)

# Calculate global average and happiness gap
global_avg = df['Ladder score'].mean()
happiest = df.loc[df['Ladder score'].idxmax()]
least_happy = df.loc[df['Ladder score'].idxmin()]
gap = happiest['Ladder score'] - least_happy['Ladder score']

print(f"Global average happiness score: {global_avg:.2f}")
print(f"Happiest country: {happiest['Country name']} "
      f"({happiest['Ladder score']:.2f})")
print(f"Least happy country: {least_happy['Country name']} "
      f"({least_happy['Ladder score']:.2f})")
print(f"Gap between the two: {gap:.2f} points")

# Step 2: Create the chart
colors = {
    'Happiest 8': '#168C78',
    'Least Happy 8': '#D95D5D'
}

fig = px.bar(
    combined,
    x='Ladder score',
    y='Country name',
    orientation='h',
    color='Group',
    color_discrete_map=colors,
    text='Ladder score',
    category_orders={
        'Country name': combined['Country name'].tolist(),
        'Group': ['Least Happy 8', 'Happiest 8']
    },
    title=(
        f"A {gap:.2f}-Point Happiness Gap Separates "
        f"{happiest['Country name']} from {least_happy['Country name']}"
        "<br><sup>Top 8 and bottom 8 countries compared with the global average</sup>"
    ),
    labels={
        'Ladder score': 'Happiness Score',
        'Country name': '',
        'Group': 'Country group'
    }
)

# Step 3: Add labels and global average reference line
fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside',
    cliponaxis=False
)

fig.add_vline(
    x=global_avg,
    line_dash='dash',
    line_color='#333333',
    line_width=2,
    annotation_text=f'Global average: {global_avg:.2f}',
    annotation_position='top'
)

# Step 4: Style the chart
fig.update_layout(
    template='plotly_white',
    title_x=0.02,
    title_font_size=19,
    xaxis=dict(
        title='Happiness Score',
        rangemode='tozero',
        showgrid=True,
        gridcolor='#EAEAEA',
        zeroline=True
    ),
    yaxis=dict(
        title='',
        categoryorder='array',
        categoryarray=combined['Country name'].tolist()
    ),
    legend_title_text='',
    height=750,
    margin=dict(l=150, r=100, t=120, b=60),
    bargap=0.25
)

fig.show(renderer='iframe')
```

```python

import pandas as pd
import plotly.express as px

# Step 1: Map countries to the five requested regions
region_mapping = {
    # Western Europe
    'Finland': 'Western Europe', 'Denmark': 'Western Europe',
    'Iceland': 'Western Europe', 'Sweden': 'Western Europe',
    'Norway': 'Western Europe', 'Netherlands': 'Western Europe',
    'Switzerland': 'Western Europe', 'Austria': 'Western Europe',
    'Belgium': 'Western Europe', 'Germany': 'Western Europe',
    'France': 'Western Europe', 'United Kingdom': 'Western Europe',
    'Ireland': 'Western Europe', 'Luxembourg': 'Western Europe',

    # Latin America
    'Brazil': 'Latin America', 'Argentina': 'Latin America',
    'Chile': 'Latin America', 'Colombia': 'Latin America',
    'Mexico': 'Latin America', 'Peru': 'Latin America',
    'Uruguay': 'Latin America', 'Ecuador': 'Latin America',
    'Costa Rica': 'Latin America', 'Panama': 'Latin America',
    'Paraguay': 'Latin America', 'Bolivia': 'Latin America',
    'Venezuela': 'Latin America', 'Guatemala': 'Latin America',
    'Honduras': 'Latin America', 'Nicaragua': 'Latin America',
    'El Salvador': 'Latin America', 'Dominican Republic': 'Latin America',

    # East Asia
    'China': 'East Asia', 'Japan': 'East Asia',
    'South Korea': 'East Asia', 'Mongolia': 'East Asia',

    # Sub-Saharan Africa
    'South Africa': 'Sub-Saharan Africa', 'Nigeria': 'Sub-Saharan Africa',
    'Kenya': 'Sub-Saharan Africa', 'Ghana': 'Sub-Saharan Africa',
    'Ethiopia': 'Sub-Saharan Africa', 'Uganda': 'Sub-Saharan Africa',
    'Tanzania': 'Sub-Saharan Africa', 'Rwanda': 'Sub-Saharan Africa',
    'Senegal': 'Sub-Saharan Africa', 'Cameroon': 'Sub-Saharan Africa',
    'Zambia': 'Sub-Saharan Africa', 'Zimbabwe': 'Sub-Saharan Africa',
    'Botswana': 'Sub-Saharan Africa', 'Namibia': 'Sub-Saharan Africa',
    'Mozambique': 'Sub-Saharan Africa', 'Malawi': 'Sub-Saharan Africa',
    'Madagascar': 'Sub-Saharan Africa', 'Mauritius': 'Sub-Saharan Africa',
    'Mali': 'Sub-Saharan Africa', 'Niger': 'Sub-Saharan Africa',
    'Burkina Faso': 'Sub-Saharan Africa', 'Benin': 'Sub-Saharan Africa',
    'Togo': 'Sub-Saharan Africa', 'Sierra Leone': 'Sub-Saharan Africa',
    'Liberia': 'Sub-Saharan Africa', 'Lesotho': 'Sub-Saharan Africa',
    'Eswatini': 'Sub-Saharan Africa', 'Burundi': 'Sub-Saharan Africa',
    'Gambia': 'Sub-Saharan Africa', 'Guinea': 'Sub-Saharan Africa',
    "Côte d'Ivoire": 'Sub-Saharan Africa',

    # South Asia
    'India': 'South Asia', 'Pakistan': 'South Asia',
    'Bangladesh': 'South Asia', 'Nepal': 'South Asia',
    'Sri Lanka': 'South Asia', 'Afghanistan': 'South Asia'
}

# Step 2: Create regional data
df_task3 = df.copy()
df_task3['Region'] = df_task3['Country name'].map(region_mapping)

# Actual columns in your World Happiness dataset
gdp_col = 'Explained by: Log GDP per capita'
freedom_col = 'Explained by: Freedom to make life choices'

# Check that columns exist
required_cols = ['Region', gdp_col, freedom_col]
missing_cols = [col for col in required_cols if col not in df_task3.columns]

if missing_cols:
    print("Missing columns:", missing_cols)
else:
    regional_avg = (
        df_task3.dropna(subset=required_cols)
        .groupby('Region')[[gdp_col, freedom_col]]
        .mean()
        .reindex([
            'Western Europe',
            'Latin America',
            'East Asia',
            'Sub-Saharan Africa',
            'South Asia'
        ])
        .dropna(how='all')
        .reset_index()
    )

    # Rename for chart labels
    regional_avg = regional_avg.rename(columns={
        gdp_col: 'GDP contribution',
        freedom_col: 'Freedom contribution'
    })

    print("Regional averages:")
    display(regional_avg)

    # Step 3: Reshape for grouped bars
    plot_data = regional_avg.melt(
        id_vars='Region',
        var_name='Factor',
        value_name='Average contribution'
    )

    # Step 4: Build grouped bar chart
    fig = px.bar(
        plot_data,
        x='Region',
        y='Average contribution',
        color='Factor',
        barmode='group',
        text='Average contribution',
        color_discrete_map={
            'GDP contribution': '#2878B5',
            'Freedom contribution': '#E58B35'
        },
        title=(
            'GDP and Freedom Contribute Differently to Happiness Across Regions'
            '<br><sup>Average explained contributions in the World Happiness Report 2023</sup>'
        ),
        labels={
            'Region': '',
            'Average contribution': 'Average explained contribution',
            'Factor': 'Happiness factor'
        }
    )

    fig.update_traces(
        texttemplate='%{text:.2f}',
        textposition='outside',
        cliponaxis=False
    )

    fig.update_layout(
        template='plotly_white',
        title_x=0.02,
        height=550,
        yaxis=dict(rangemode='tozero', showgrid=True),
        legend_title_text='',
        margin=dict(t=110, b=100)
    )

    fig.show(renderer='iframe')
```
